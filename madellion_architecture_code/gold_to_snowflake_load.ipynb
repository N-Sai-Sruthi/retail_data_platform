```python
%pip install snowflake-connector-python
%pip install python-dotenv

import os
import snowflake.connector
from dotenv import load_dotenv

load_dotenv()


# -----------------------------
# Snowflake CONFIG
# -----------------------------
USER = os.getenv("SNOWFLAKE_USER")
PASSWORD = os.getenv("SNOWFLAKE_PASSWORD")
ACCOUNT = os.getenv("SNOWFLAKE_ACCOUNT")
WAREHOUSE = os.getenv("SNOWFLAKE_WAREHOUSE")
DATABASE = os.getenv("SNOWFLAKE_DATABASE")
SCHEMA = os.getenv("SNOWFLAKE_SCHEMA")

# -----------------------------
# Volume BASE PATH
# -----------------------------
BASE_PATH = "/Volumes/workspace/default/retail_data_project_capstone/"

TABLE_PATHS = {
    "dim_customer": BASE_PATH + "dim_customer/",
    "dim_product": BASE_PATH + "dim_product/",
    "dim_date": BASE_PATH + "dim_date/",
    "fact_sales": BASE_PATH + "fact_sales/"
}

# -----------------------------
# CONNECT
# -----------------------------
conn = snowflake.connector.connect(
    user=USER,
    password=PASSWORD,
    account=ACCOUNT,
    warehouse=WAREHOUSE,
    database=DATABASE,
    schema=SCHEMA
)

cur = conn.cursor()

try:
    for table, folder in TABLE_PATHS.items():
        print(f"\n🚀 Processing {table}")

        # Step 1: find parquet file
        files = [f for f in os.listdir(folder) if f.startswith("part-") and f.endswith(".parquet")]
        if not files:
            raise Exception(f"No parquet file found in {folder}")

        full_path = os.path.join(folder, files[0])

        # Step 2: truncate table (important)
        cur.execute(f"TRUNCATE TABLE {table}")

        # Step 3: upload to stage
        cur.execute(f"PUT file://{full_path} @%{table} AUTO_COMPRESS=TRUE")

        # Step 4: load data
        cur.execute(f"""
            COPY INTO {table}
            FROM @%{table}
            FILE_FORMAT = (TYPE = PARQUET)
            MATCH_BY_COLUMN_NAME = CASE_INSENSITIVE
        """)

        print(f"✅ Loaded {table}")

finally:
    cur.close()
    conn.close()
```
